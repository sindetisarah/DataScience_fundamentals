```python
#abstraction
from abc import ABC, abstractmethod

#Base class/Parent class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
  #subclass properties
    def __init__(self, radius):
        self.radius = radius
  #define a method area in our sub class
    def area(self):
        return 3.14 * self.radius ** 2
    def perimeter(self):
        return 2 * 3.14 * self.radius
#defining my object
my_circle = Circle(5)
print(my_circle.area())

#find perimeter
print(my_circle.perimeter())
```

```python
#Method Overiding
class Animal:
    def make_sound(self):
        print("The animal is making a sound")
class Dog(Animal):
    def make_sound(self):
        print("The dog is barking")

#create an object of the dog class
my_dog = Dog()
my_dog.make_sound()
```

```python
#example 2 on astraction
from abc import ABC, abstractmethod
class Vehicle():
    @abstractmethod
    def start(self):
        pass
    @abstractmethod
    def stop(self):
        pass
class Car(Vehicle):
    def start(self):
        print("Car engine started")
    def stop(self):
        print("Car engine stopped")
class Motorcycle(Vehicle):
    def start(self):
        print("Motorcycle engine started")
    def stop(self):
        print("Motorcycle engine stopped")
#instances from our above classes
#defining our object inheriting start and stop method by a MUST
my_car = Car()
my_car.start()
my_car.stop()

my_motorcycle = Motorcycle()
my_motorcycle.start()
my_motorcycle.stop()
```

```python
_#polymorphism
class Vehicle:
    def __init__(self, make, num_wheels, weight):
        self.make = make
        self.num_wheels = num_wheels
        self.weight = weight
    def start_engine(self):
        print(f"The {self.make} engine is starting.")

    #subclass Car
class Car(Vehicle):
    def __init__(self, make, num_wheels, weight, num_doors):
        super().__init__(make, num_wheels, weight)
        self.num_doors = num_doors

    def start_engine(self):
        print(f"The {self.make} car engine is starting.")
    #subclass boot
class Boat(Vehicle):
    def __init__(self, make, num_wheels, weight, boot_size):
        super().__init__(make, num_wheels, weight)
        self.boot_size = boot_size
    def start_engine(self):
        print(f"The {self.make} boat engine is starting.")

#creating an object and applying the above methods
sarahTesla = Vehicle("Tesla", 4, 2000)
sarahTesla.start_engine()
#invoking the methods of the sub classes
sarahsToyota = Car("Toyota", 4, 2000, 4)
sarahsToyota.start_engine()
#invoking the method on Boot Class
sarahsBoat = Boat("FerryBoat", 4, 2000, 4)
sarahsBoat.start_engine()
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def get_balance(self):
        return self.balance
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. Your new balance is {self.balance}")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"You have Withdrew {amount}. Your balance is {self.balance}")
        else:
            print("Insufficient funds")

account = BankAccount(100000)
print(f"Your account balance is {account.get_balance()}")
account.deposit(50000)
print(account.get_balance())
print(account.get_balance())
```

```python
#inheritance
class Vehicle:
    def __init__(self, make, year):
        self.make = make
        self.year = year
    def start_engine(self):
        print(f"The {self.year} {self.make} engine is starting.")
    def stop_engine(self):
        print(f"The {self.year} {self.make} engine is stopping.")
class Car(Vehicle):
    def __init__(self, make, year, color):
           super().__init__(make, year)
           self.color = color

    def open_trunk(self):
        print(f"My {self.color} {self.make} is the latest a {self.year} lets check the trunk opening.")

sarahsCar = Car("Tesla", 2025, "Red") # Include color when creating the Car object
sarahsCar.start_engine()
sarahsCar.open_trunk()
sarahsCar.stop_engine()
```

```python
#Object Oriented Programming
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print(f"The {self.year} {self.make} {self.model}'s engine is starting.")

    def drive(self):
        print(f"The {self.year} {self.make} {self.model} is driving.")

sindetsCar = Car("Toyota", "Camry", 2025)
sindetsCar.start_engine()
sindetsCar.drive()
```
