```python
#decorators functions uppercase
def uppercase_decorator(function):
  def wrapper():
    result = function()
    return result.upper()
  return wrapper

@uppercase_decorator
def greet():
  return "hello world"

print(greet())
```

```python
#generator function to genearate squares
def square_numbers(n):
  for i in range(n):
    yield i ** 2
squares = square_numbers(5)
for square in squares:
  print(square)
```

```python
#generators
numbers = [1, 2, 3, 4, 5]
for number in numbers:
  print(number ** 2)
```

```python
#reduce function
from functools import reduce
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x + y, numbers)
print(product)
```

```python
#even numbers using lambda
numbers=range(1,20)
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

```python
#even numbers using lambda
even_numbers = list(filter(lambda x: x % 2 == 0, range(1, 11)))
print(even_numbers)
```

```python
#convert a list of temperatures in Celsius to Fahrenheit using map. Formula: F = (C * 9/5) + 32.
celsius_temperatures = [25, 30, 15, 20]
fahrenheit_temperatures = list(map(lambda c: (c * 9/5) + 32, celsius_temperatures))
print(fahrenheit_temperatures)
```

```python
x = 10 ;
if x > 0 :
  print("x is positive")

def predictPositive(x) :
  if x > 0 :
    print("x is positive")
  else :
    print("x is not positive")


predictPositive(10)
predictPositive(-10)
```

```python
#Control Flow Structures in Python
weather ="rainy"
if weather == "rainy":
  print("Carry an umbrella")
else:
  print("Don't carry an umbrella")

def predictWeather(weather) :
  if weather == "rainy":
    print("Carry an umbrella")
  else:
    print("Don't carry an umbrella")

predictWeather("sunny")
predictWeather("rainy")
```
